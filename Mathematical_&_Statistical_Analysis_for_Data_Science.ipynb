###Task 1 – Linear Algebra Implementation

● Perform vector addition, subtraction, and dot product using vector_a and vector_b

● Perform matrix addition and matrix multiplication using matrix_a and matrix_b

● Calculate determinant of matrix_a

● If determinant of matrix_a is not equal to zero, then find inverse of matrix_a

● Compute eigenvalues and eigenvectors of matrix_a

```python
# dataset

import numpy as np
vector_a = np.array([2, 4, 6])
vector_b = np.array([1, 3, 5])
matrix_a = np.array([
[2, 1, 3],
[1, 0, 2],
[4, 1, 1]
])
matrix_b = np.array([
[1, 2, 0],
[3, 1, 4],
[2, 5, 1]
])
```

```python
print("vector_a :", vector_a)
print("vector_b :", vector_b)
```

```python
 # Perform vector addition, subtraction, and dot product using vector_a and vector_b

vec_add = vector_a + vector_b
vec_sub = vector_a - vector_b
vec_dot = np.dot(vector_a, vector_b)

print("Vector Addition :", vec_add)
print("Vector Subtraction :", vec_sub)
print("Vector Dot Product :", vec_dot)
```

```python
print("matrix_a :\n", matrix_a)
print("matrix_b :\n", matrix_b)
```

```python
# Perform matrix addition and matrix multiplication using matrix_a and matrix_b

mat_add = matrix_a + matrix_b
mat_mul = np.dot(matrix_a, matrix_b)

print("Matrix Addition :\n", mat_add)
print("Matrix Multiplication :\n", mat_mul)
```

```python
# Calculate determinant of matrix_a

det_matrix_a = np.linalg.det(matrix_a)

print("Determinant of matrix_a :", det_matrix_a)
```

```python
# If determinant of matrix_a is not equal to zero, then find inverse of matrix_a

det_matrix_a = np.linalg.det(matrix_a)

if det_matrix_a != 0:
    inverse_matrix_a = np.linalg.inv(matrix_a)
    print("Inverse of matrix_a :\n", inverse_matrix_a)
    print("Determinant of matrix_a is not equal to zero")
else:
    print("Determinant of matrix_a is equal to zero")
```

```python
# Compute eigenvalues and eigenvectors of matrix_a

eigenvalues, eigenvectors = np.linalg.eig(matrix_a)

print("Eigenvalues of matrix_a :", eigenvalues)
print("Eigenvectors of matrix_a :\n", eigenvectors)
```

### Task 2 – Probability Concepts

####Using Python / NumPy / SciPy:
● Calculate probability of successful sales

● Demonstrate conditional probability using customer segments

```python
# Dataset

sales_success = [0.65, 0.35]
customer_segments = ["Premium", "Regular"]
segment_purchase_probability = {
"Premium": 0.80,
"Regular": 0.50
}

sales_success
```

```python
# Calculate probability of successful sales

total_sales = sum(sales_success)
probability_success = [success / total_sales for success in sales_success]

print("Probability of Successful Sales:")
for i in range(len(customer_segments)):
    print(f"{customer_segments[i]}: {probability_success[i]}")
```

```python
print("Or :")

# Probability of successful sales
prob_success = sales_success[0]

print("Probability of successful sales:", prob_success)
print("Successful sales percentage:", prob_success * 100, "%")
```

```python
# Demonstrate conditional probability using customer_segments

import numpy as np
from scipy import stats


print("\nConditional Probability:")

# Premium customers
premium = segment_purchase_probability["Premium"]
print("P(Purchase | Premium):", premium)
print("Premium purchase percentage:", premium * 100, "%")

# Regular customers
regular = segment_purchase_probability["Regular"]
print("P(Purchase | Regular):", regular)
print("Regular purchase percentage:", regular * 100, "%")
```

###Bayes’ theorem :

● Apply Bayes’ theorem to calculate updated probability of a customer purchase

○ Formula: P(A|B) = [P(B|A) * P(A)] / P(B)

```python
# Data for Bayes’ theorem :

# Probability Dataset for Customer Purchase Analysis

# P(A): Probability that a customer is Premium
P_A = 0.40
# P(B): Probability that a customer makes a purchase
P_B = 0.62
# P(B|A): Probability that a Premium customer makes a purchase
P_B_given_A = 0.80

# P(A|B) = [P(B|A) * P(A)] / P(B)

P_A_given_B = (P_B_given_A * P_A) / P_B

print("Updated Probability of a Customer Purchase (P(A|B)):", P_A_given_B)
print("Updated purchase percentage:", P_A_given_B * 100, "%")
```

### Task 3 – Descriptive Statistics

● Compute mean, median, and mode of sales_data

● Calculate range, variance, and standard deviation

● Measure skewness and kurtosis

● Interpret whether the dataset is symmetric, positively skewed, or negatively skewed

```python
# Statistical Dataset

sales_data = [120, 135, 150, 145, 160, 155, 170, 165, 180, 175,
190, 200, 210, 195, 185, 175, 165, 155, 145, 135]

print(sales_data)
```

```python
# Compute mean, median, and mode of sales_data

mean_sales = np.mean(sales_data)
median_sales = np.median(sales_data)
mode_sales = stats.mode(sales_data)

print("Mean of sales_data:", mean_sales)
print("Median of sales_data:", median_sales)
print("Mode of sales_data:", mode_sales)
```

```python
# Calculate range, variance, and standard deviation

range_sales = np.max(sales_data) - np.min(sales_data)
variance_sales = np.var(sales_data)
std_sales = np.std(sales_data)

print("Range of sales_data:", range_sales)
print("Variance of sales_data:", variance_sales)
print("Standard Deviation of sales_data:", std_sales)
```

```python
# Measure skewness and kurtosis

skewness_sales = stats.skew(sales_data)
kurtosis_sales = stats.kurtosis(sales_data)

print("Skewness of sales_data:", skewness_sales)
print("Kurtosis of sales_data:", kurtosis_sales)
```

```python
# Using NumPy / Pandas / statistics library:

# Interpret whether the dataset is symmetric, positively skewed, or negatively skewed

symmetrickness_sales = stats.skew(sales_data)
positive_skewness = symmetrickness_sales > 0
negative_skewness = symmetrickness_sales < 0

print("Symmetricity of sales_data:", symmetrickness_sales)
print("Is sales_data positively skewed?", positive_skewness)
print("Is sales_data negatively skewed?", negative_skewness)
```
